# 4. Data Preprocessing & Cleaning


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,roc_auc_score)
from sklearn.model_selection import GridSearchCV
import warnings
warnings.filterwarnings("ignore")


Data=pd.read_csv("online_shoppers_intention.csv")
print(Data.head())
```

```python
sns.boxplot(x=Data['ProductRelated_Duration'])
plt.title("ProductRelated_Duration Boxplot")
plt.show()
```

```python
Q1 = Data['ProductRelated_Duration'].quantile(0.25)
Q3 = Data['ProductRelated_Duration'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

Data = Data[(Data['ProductRelated_Duration'] >= lower_bound) &
            (Data['ProductRelated_Duration'] <= upper_bound)]
```

# 6. Feature Engineering

```python
Data.select_dtypes(include=['object']).columns
```

```python
le = LabelEncoder()

Data['Month'] = le.fit_transform(Data['Month'])
Data['VisitorType'] = le.fit_transform(Data['VisitorType'])
```

```python
Data['Weekend'] = Data['Weekend'].astype(int)
Data['Revenue'] = Data['Revenue'].astype(int)
```

```python
Data.info()
```

# 7. Feature Scaling

```python
X = Data.drop('Revenue', axis=1)
y = Data['Revenue']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(X_scaled, columns=X.columns)
X_scaled.head()
```

# 8. Model Building

```python

X = Data.drop("Revenue", axis=1)
y = Data["Revenue"]

X_train,X_test,y_train, y_test = train_test_split(X_scaled,y,test_size=0.2,random_state=42)
```

<h2>1-Logistic Regression</h2>

```python
lr = LogisticRegression()
lr.fit(X_train, y_train)
```

```python
y_pred_lr = lr.predict(X_test)
```

<h2>2-Decision Tree</h2>

```python
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
```

```python
y_pred_dt = dt.predict(X_test)
```

<h2>3-Random Forest</h2>

```python
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
```

```python
import pickle
pickle.dump(rf,open("model.pkl","wb"))
pickle.dump(scaler,open("scaler.pkl","wb"))
```

```python
y_pred_rf = rf.predict(X_test)
```

# 9-Model Evaluation

<h5>Logistic Regression</h5>

```python
print("Logistic Regression\n")
print("Accuracy :", accuracy_score(y_test, y_pred_lr))
print("Precision:", precision_score(y_test, y_pred_lr))
print("Recall   :", recall_score(y_test, y_pred_lr))
print("F1 Score :", f1_score(y_test, y_pred_lr))
print("\nConfusion_Matrix")
confusion_matrix(y_test, y_pred_lr)
```

```python
print("\nROC-AUC\n")
roc_auc = roc_auc_score(y_test, lr.predict_proba(X_test)[:,1])
print("ROC-AUC Score:", roc_auc)
```

<h5>Decision Tree</h5>

```python
print("Decision Tree\n")

print("Accuracy :", accuracy_score(y_test, y_pred_dt))
print("Precision:", precision_score(y_test, y_pred_dt))
print("Recall   :", recall_score(y_test, y_pred_dt))
print("F1 Score :", f1_score(y_test, y_pred_dt))
print("\nConfusion_Matrix")
confusion_matrix(y_test, y_pred_dt)
```

```python
print("\nROC-AUC\n")
roc_auc = roc_auc_score(y_test, dt.predict_proba(X_test)[:,1])
print("ROC-AUC Score:", roc_auc)
```

<h5>Random Forest</h5>

```python
print("Random Forest\n")

print("Accuracy :", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall   :", recall_score(y_test, y_pred_rf))
print("F1 Score :", f1_score(y_test, y_pred_rf))
print("\nConfusion_Matrix")
confusion_matrix(y_test, y_pred_rf)
```

```python
print("\nROC-AUC\n")
roc_auc = roc_auc_score(y_test, rf.predict_proba(X_test)[:,1])
print("ROC-AUC Score:", roc_auc)
```

<h4>Several classification models were evaluated using Accuracy, Precision, Recall, F1-Score, Confusion Matrix, and ROC-AUC Score. The Random Forest Classifier achieved the highest accuracy of 92% and showed the best overall performance. Therefore, it was selected as the final model for predicting customer purchase behavior and used in the web application.</h4>

# 10. Hyperparameter Tuning

```python
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}
```

```python
grid_search = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(grid_search.best_params_)
print("Best Cross Validation Accuracy:")
print(grid_search.best_score_)
```

```python
best_rf = grid_search.best_estimator_
y_pred = best_rf.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```
